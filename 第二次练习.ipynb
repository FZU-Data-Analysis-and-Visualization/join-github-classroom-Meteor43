```python
arr = np.array([10, 15, 20, 25, 30, 35])
# 1. 花式索引提取第1、3、5个元素（索引0,2,4）
fancy_index = arr[[0, 2, 4]]
# 2. 布尔索引提取大于20的元素
bool_index = arr[arr > 20]
print("原数组：", arr)
print("花式索引结果：", fancy_index)
print("布尔索引结果：", bool_index)
```

Q6

```python
import numpy as np
# 1. 创建数组
x = np.array([1, 4, 9, 16])
y = np.array([2, 1, 4, 5])
# 2. 计算数组 x 中每个元素的平方根
sqrt_x = np.sqrt(x)
# 3. 计算对应位置上的最大值
max_result = np.maximum(sqrt_x, y)
print("数组 x：", x)
print("数组 y：", y)
print("x 的平方根：", sqrt_x)
print("对应位置的最大值：", max_result)
```

```python
import numpy as np
# 创建形状为(2, 3, 4)的三维数组
arr = np.arange(24).reshape((2, 3, 4))
# 使用transpose()转换轴顺序为(1, 0, 2)
transposed_arr = arr.transpose((1, 0, 2))
# 打印转换后的形状
print("原数组形状：", arr.shape)
print("转置后数组形状：", transposed_arr.shape)
```

Q8

```python
import numpy as np
# 创建题目给定的数组
arr_x = np.array([1, 5, 7])
arr_y = np.array([2, 6, 8])
arr_con = np.array([True, False, True])
# 使用 np.where() 进行条件选择
result = np.where(arr_con, arr_x, arr_y)
print("条件数组：", arr_con)
print("arr_x：", arr_x)
print("arr_y：", arr_y)
print("结果数组：", result)
```

Q9

```python
import numpy as np
# 创建二维数组
arr = np.array([[6, 2, 7], [3, 6, 2], [4, 3, 2]])
# 1. 计算所有元素的算术平均值
mean_val = arr.mean()
# 2. 按列(axis=0)从小到大排序
arr_sorted = np.sort(arr, axis=0)
print("原数组：")
print(arr)
print("所有元素的算术平均值：", mean_val)
print("按列排序后的数组：")
print(arr_sorted)
```
