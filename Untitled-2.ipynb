```python
import pandas as pd
# 创建含缺失值的 DataFrame
data = {"x": [1, None, 3], "y": [None, 5, 6]}
df_missing = pd.DataFrame(data)
# 检测缺失值并统计每列缺失数量
print("缺失值检测结果：")
print(df_missing.isnull())  # 显示每个位置是否为缺失值
print("\n每列缺失值数量：")
print(df_missing.isnull().sum())  # 统计每列缺失值个数
```

3.2

```python
# 用 0 填充缺失值
df_fill = df_missing.fillna(0)
print("用 0 填充后的 DataFrame：")
print(df_fill)

# 删除含缺失值的行
df_drop = df_missing.dropna()
print("\n删除含缺失值行后的 DataFrame：")
print(df_drop)
```

4.1

```python
# 按 gender 分组，同时计算均值和最大值
group_result = df1.groupby("gender")["score"].agg(["mean", "max"])
print("按 gender 分组的 score 均值和最大值：")
print(group_result)
```

4.2

```python
# 方法1：value_counts() 直接统计
count_result = df1["gender"].value_counts()
print("按 gender 分组的人数统计：")
print(count_result)

# 方法2：groupby + count()
count_result2 = df1.groupby("gender").size()
print("\n按 gender 分组的人数统计：")
print(count_result2)
```

5.1

```python
df_sorted = df1.sort_values(by='score', ascending=True)
print("按 score 升序排序后的 DataFrame：")
print(df_sorted)
```

5.2

```python
# 添加一行重复数据（Tom,85,男）
new_row = pd.DataFrame([{"name":"Tom", "score":85, "gender":"男","grade":"良好"}])
df_duplicate = pd.concat([df1, new_row], ignore_index=True)
print("添加重复数据后的 DataFrame：")
print(df_duplicate)

# 去重（默认保留第一次出现的行）
df_deduped = df_duplicate.drop_duplicates()
print("\n去重后的 DataFrame：")
print(df_deduped)
```

```python
df1 = pd.DataFrame({"id":[1,2], "name":["X","Y"]})
df2 = pd.DataFrame({"id":[1,2], "age":[20,30]})

# 按 id 内连接合并（保留两边共有的 id）
df_merged = pd.merge(df1, df2, on='id', how='inner')
print("横向合并后的 DataFrame：")
print(df_merged)
```

```python
import pandas as pd

# 创建时间列表
dates = ["2025-01-01", "2025-01-02", "2025-01-03"]
df_time = pd.DataFrame({"date": dates})

# 转为 datetime 格式
df_time['date'] = pd.to_datetime(df_time['date'])
print("转换后的时间列：")
print(df_time.dtypes)
print(df_time)
```

7.2

```python
# 提取月份
df_time['month'] = df_time['date'].dt.month
# 提取星期（0=周一，6=周日；也可用dt.day_name()获取星期名称）
df_time['weekday_num'] = df_time['date'].dt.weekday
df_time['weekday_name'] = df_time['date'].dt.day_name()

print("\n提取月份和星期后的 DataFrame：")
print(df_time)
```

8.1

```python
score_mean = df_sorted['score'].mean()
score_std = df_sorted['score'].std()

# 标准化
df_sorted['score_standardized'] = (df_sorted['score'] - score_mean) / score_std

print("标准化后的 score 列：")
print(df_sorted[['score', 'score_standardized']])
```

8.2

```python
# 使用 pivot_table
pivot_result = df_sorted.pivot_table(
    index='gender', 
    values='score', 
    aggfunc='median'
)

print("\n按 gender 分组的 score 中位数透视表：")
print(pivot_result)
```
